```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

train_data = pd.read_csv('../data/processed/train.csv')
val_data = pd.read_csv('../data/processed/val.csv')
test_data = pd.read_csv('../data/processed/test.csv')

# Выделяем признаки (X) и целевую переменную (y)
X_train = train_data.drop('LEAVE', axis=1)
y_train = train_data['LEAVE']

X_val = val_data.drop('LEAVE', axis=1)
y_val = val_data['LEAVE']

X_test = test_data.drop('LEAVE', axis=1)
y_test = test_data['LEAVE']

print(f"Размер X_train: {X_train.shape}")
print(f"Размер X_val: {X_val.shape}")
print(f"Размер X_test: {X_test.shape}")
```

```python
# Список признаков, которые создал сам (Удаление Feature Engineering)
engineered_features = ['Income_per_House', 'Has_Overage', 'Total_Long_Calls_Duration']

# Удаляем их из всех выборок
X_train_base = X_train.drop(columns=engineered_features)
X_val_base = X_val.drop(columns=engineered_features)
X_test_base = X_test.drop(columns=engineered_features)

print(f"Количество признаков для Baseline: {X_train_base.shape[1]}")
```

```python
baseline_model = LogisticRegression(random_state=42)
baseline_model.fit(X_train_base, y_train)
print("Модель успешно обучена!")
```

```python
def evaluate_model(model, X, y_true, dataset_name):
    # Предсказание жестких классов (0 или 1) для F1, Accuracy, Precision, Recall
    y_pred = model.predict(X)
    # Предсказание вероятностей (от 0 до 1) для ROC-AUC (берем вероятности класса 1)
    y_pred_proba = model.predict_proba(X)[:, 1]
    
    print(f"--- Метрики на выборке: {dataset_name} ---")
    print(f"ROC-AUC:   {roc_auc_score(y_true, y_pred_proba):.4f}")
    print(f"F1-score:  {f1_score(y_true, y_pred):.4f}")
    print(f"Accuracy:  {accuracy_score(y_true, y_pred):.4f}")
    print(f"Precision: {precision_score(y_true, y_pred):.4f}")
    print(f"Recall:    {recall_score(y_true, y_pred):.4f}")

evaluate_model(baseline_model, X_val_base, y_val, "VALIDATION")
evaluate_model(baseline_model, X_test_base, y_test, "TEST")
```
